```python
import pandas as pd
import matplotlib.pyplot as plt

# Import Dataset
df = pd.read_csv("ev_sales_by_makers_and_cat_15-24.csv")


# Dataset Inspection
print("===== FIRST FIVE RECORDS =====")
display(df.head())

print("\n===== SHAPE OF DATASET =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SUMMARY =====")
display(df.describe())
```

```python
# Filtering Data

# EV makers with 2024 sales greater than 1000
print("\n===== EV MAKERS WITH 2024 SALES > 1000 =====")
high_sales = df[df["2024"] > 1000]
display(high_sales.head())

# 2-Wheeler EV bookings
print("\n===== 2-WHEELER EV SALES =====")
two_wheeler = df[df["Cat"] == "2W"]
display(two_wheeler.head())
```

```python
# Statistical Analysis

print("\n===== 2024 EV SALES STATISTICS =====")
print("Mean :", df["2024"].mean())
print("Median :", df["2024"].median())
print("Mode :")
print(df["2024"].mode())
print("Standard Deviation :", df["2024"].std())

print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())
```

```python
# Grouping and Aggregation

print("\n===== AVERAGE 2024 EV SALES BY VEHICLE CATEGORY =====")
category_avg = df.groupby("Cat")["2024"].mean()
display(category_avg)
```

```python
# Correlation Analysis

print("\n===== CORRELATION =====")

corr = df[[
    "2015",
    "2016",
    "2017",
    "2018",
    "2019",
    "2020",
    "2021",
    "2022",
    "2023",
    "2024"
]].corr()

display(corr)
```

```python
# Visualization 1

category_avg.plot(kind="bar")

plt.title("Average 2024 EV Sales by Vehicle Category")
plt.xlabel("Vehicle Category")
plt.ylabel("Average EV Sales")

plt.show()
```

```python
# Visualization 2

plt.figure(figsize=(8,5))

plt.scatter(df["2023"], df["2024"])

plt.title("2023 EV Sales vs 2024 EV Sales")

plt.xlabel("EV Sales in 2023")

plt.ylabel("EV Sales in 2024")

plt.show()
```
