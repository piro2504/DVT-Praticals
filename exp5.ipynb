```python
# ============================================================
# LOAD DATASET - VS CODE
# ============================================================

df = pd.read_csv("ev_sales_by_makers_and_cat_15-24.csv")

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# ============================================================
# DATA TYPES
# ============================================================

print("\nData Types:")

print(df.dtypes)
```

```python
# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics:")

display(df.describe())
```

```python
# ============================================================
# MISSING VALUES
# ============================================================

print("\nMissing Values:")

print(df.isnull().sum())
```

```python
# ============================================================
# DUPLICATE ROWS
# ============================================================

print("\nDuplicate Rows:")

print(df.duplicated().sum())
```

```python
# ============================================================
# DATA CLEANING
# ============================================================

# Remove duplicate rows
df = df.drop_duplicates()

# Identify year columns
year_cols = [
    '2015', '2016', '2017', '2018', '2019',
    '2020', '2021', '2022', '2023', '2024'
]

# Convert year columns to numeric
for col in year_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Fill missing numerical values with 0
df[year_cols] = df[year_cols].fillna(0)

print("\nMissing Values After Cleaning:")

print(df.isnull().sum())
```

```python
# ============================================================
# TOTAL EV SALES
# ============================================================

df['Total_Sales'] = df[year_cols].sum(axis=1)

print("\nTotal EV Sales calculated successfully.")

display(
    df[['Cat', 'Maker', 'Total_Sales']]
    .sort_values(by='Total_Sales', ascending=False)
    .head(10)
)
```

```python
# ============================================================
# FIND NUMERIC COLUMNS
# ============================================================

numeric_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

print("\nNumeric Columns:")

print(numeric_cols)
```

```python
# ============================================================
# FIND CATEGORICAL COLUMNS
# ============================================================

categorical_cols = df.select_dtypes(
    include="object"
).columns.tolist()

print("\nCategorical Columns:")

print(categorical_cols)
```

```python
# ============================================================
# EV SALES: 2023 VS 2024 SCATTER PLOT
# ============================================================

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="2023",
    y="2024"
)

plt.title("EV Sales: 2023 vs 2024")

plt.xlabel("EV Sales in 2023")

plt.ylabel("EV Sales in 2024")

plt.show()
```

```python
# ============================================================
# TOTAL EV SALES OUTLIER DETECTION
# ============================================================

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Total_Sales"]
)

plt.title("EV Total Sales Outlier Detection")

plt.xlabel("Total EV Sales")

plt.show()
```

```python
# ============================================================
# 2024 EV SALES DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["2024"]
)

plt.title("EV Sales Distribution in 2024")

plt.xlabel("EV Sales in 2024")

plt.show()
```

```python
# ============================================================
# CORRELATION HEATMAP
# ============================================================

corr = df[year_cols].corr()

print("\nCorrelation Matrix:")

display(corr)

plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("EV Sales Correlation Heatmap")

plt.show()
```

```python
# ============================================================
# EV CATEGORY COUNT
# ============================================================

plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="Cat",
    order=df["Cat"].value_counts().index
)

plt.xticks(rotation=45)

plt.title("Electric Vehicle Categories")

plt.xlabel("EV Category")

plt.ylabel("Number of Records")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# TOP 10 EV MAKERS
# ============================================================

top_makers = (
    df["Maker"]
    .value_counts()
    .head(10)
    .index
)

plt.figure(figsize=(10,5))

sns.countplot(
    data=df[
        df["Maker"].isin(top_makers)
    ],
    x="Maker",
    order=top_makers
)

plt.xticks(rotation=45)

plt.title("Top 10 EV Makers")

plt.xlabel("EV Maker")

plt.ylabel("Number of Records")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# TOTAL EV SALES DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))

sns.histplot(
    df["Total_Sales"],
    bins=30,
    kde=True
)

plt.title(
    "Electric Vehicle Total Sales Distribution"
)

plt.xlabel("Total EV Sales")

plt.ylabel("Number of Records")

plt.show()
```

```python
# ============================================================
# 2024 EV SALES DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))

sns.histplot(
    df["2024"],
    bins=20,
    kde=True
)

plt.title(
    "Electric Vehicle Sales Distribution - 2024"
)

plt.xlabel("EV Sales in 2024")

plt.ylabel("Number of Records")

plt.show()
```

```python
# ============================================================
# EV CATEGORY DISTRIBUTION - PIE CHART
# ============================================================

category = df["Cat"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    category,
    labels=category.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title(
    "Electric Vehicle Category Distribution"
)

plt.show()
```

```python
# ============================================================
# AVERAGE TOTAL SALES BY EV CATEGORY
# ============================================================

category_sales = (
    df.groupby("Cat")["Total_Sales"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

category_sales.plot(
    kind="bar"
)

plt.title(
    "Average EV Sales by Category"
)

plt.xlabel("EV Category")

plt.ylabel("Average Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# AVERAGE 2024 SALES BY EV CATEGORY
# ============================================================

category_2024 = (
    df.groupby("Cat")["2024"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

category_2024.plot(
    kind="bar"
)

plt.title(
    "Average EV Sales by Category - 2024"
)

plt.xlabel("EV Category")

plt.ylabel("Average 2024 Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# TOP 10 EV MAKERS BY AVERAGE TOTAL SALES
# ============================================================

maker_sales = (
    df.groupby("Maker")["Total_Sales"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

maker_sales.plot(
    kind="bar"
)

plt.title(
    "Top 10 EV Makers by Average Total Sales"
)

plt.xlabel("EV Maker")

plt.ylabel("Average Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# EV SALES GROWTH BY CATEGORY
# ============================================================

category_year = (
    df.groupby("Cat")[["2023", "2024"]]
    .sum()
)

category_year.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title(
    "EV Sales Comparison by Category: 2023 vs 2024"
)

plt.xlabel("EV Category")

plt.ylabel("Total EV Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# 2023 VS 2024 EV SALES BY TOP MAKERS
# ============================================================

top_maker_sales = (
    df.groupby("Maker")[["2023", "2024"]]
    .sum()
    .sort_values("2024", ascending=False)
    .head(10)
)

top_maker_sales.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title(
    "Top 10 EV Makers: 2023 vs 2024 Sales"
)

plt.xlabel("EV Maker")

plt.ylabel("EV Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# EV SALES BUBBLE CHART
# ============================================================

plt.figure(figsize=(9,6))

bubble_size = (
    df["Total_Sales"]
    .fillna(df["Total_Sales"].median())
    / df["Total_Sales"].max()
    * 500
)

plt.scatter(
    df["2015"],
    df["2024"],
    s=bubble_size,
    alpha=0.5,
    c=df["2024"],
    cmap="viridis"
)

plt.colorbar(
    label="2024 EV Sales"
)

plt.xlabel("EV Sales in 2015")

plt.ylabel("EV Sales in 2024")

plt.title(
    "Bubble Chart: EV Sales Growth from 2015 to 2024"
)

plt.show()
```

```python
# ============================================================
# CATEGORY VS TOTAL SALES - VIOLIN PLOT
# ============================================================

plt.figure(figsize=(12,6))

sns.violinplot(
    data=df,
    x="Cat",
    y="Total_Sales"
)

plt.xticks(rotation=45)

plt.title(
    "Total EV Sales Distribution by Category"
)

plt.xlabel("EV Category")

plt.ylabel("Total EV Sales")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CATEGORY VS 2024 SALES - VIOLIN PLOT
# ============================================================

plt.figure(figsize=(12,6))

sns.violinplot(
    data=df,
    x="Cat",
    y="2024"
)

plt.xticks(rotation=45)

plt.title(
    "2024 EV Sales Distribution by Category"
)

plt.xlabel("EV Category")

plt.ylabel("2024 EV Sales")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CATEGORY VS TOTAL SALES - BOX PLOT
# ============================================================

plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="Cat",
    y="Total_Sales"
)

plt.xticks(rotation=45)

plt.title(
    "Total EV Sales by Category"
)

plt.xlabel("EV Category")

plt.ylabel("Total EV Sales")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# PAIRPLOT
# ============================================================

pair_columns = [
    "2015",
    "2018",
    "2020",
    "2022",
    "2024"
]

sns.pairplot(
    df[pair_columns].dropna()
)

plt.show()
```

```python
# ============================================================
# YEAR-WISE EV ADOPTION TREND
# ============================================================

year_sales = df[year_cols].sum()

plt.figure(figsize=(12,6))

plt.plot(
    year_sales.index,
    year_sales.values,
    marker='o',
    linewidth=3
)

plt.title(
    "Electric Vehicle Adoption Trend: 2015-2024"
)

plt.xlabel("Year")

plt.ylabel("Total EV Sales")

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CATEGORY-WISE EV ADOPTION TREND
# ============================================================

category_trend = df.groupby("Cat")[year_cols].sum()

plt.figure(figsize=(12,7))

for category in category_trend.index:

    plt.plot(
        year_cols,
        category_trend.loc[category],
        marker='o',
        label=category
    )

plt.title(
    "EV Adoption Trend by Vehicle Category"
)

plt.xlabel("Year")

plt.ylabel("EV Sales")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()
```
