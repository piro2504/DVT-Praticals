```python
import pandas as pd
import numpy as np
df = pd.read_csv("ev_sales_by_makers_and_cat_15-24.csv")

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate rows after removal:", df.duplicated().sum())

print(df.head())
```

```python
Q1 = df["2024"].quantile(0.25)

Q3 = df["2024"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[(df["2024"] < lower) | (df["2024"] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

num_cols = df.columns[2:]

df[num_cols] = scaler.fit_transform(df[num_cols])

print(df.head())
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Maker"] = encoder.fit_transform(df["Maker"])

df["Cat"] = encoder.fit_transform(df["Cat"])

print(df.head())
```

```python
df = pd.get_dummies(df,
                    columns=["Cat"],
                    dtype=int)

print(df.head())
```
