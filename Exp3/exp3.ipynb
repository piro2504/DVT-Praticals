```python
# EX NO: 3 - Data Cleaning


import pandas as pd

# Load dataset
df = pd.read_csv("ev_sales_by_makers_and_cat_15-24.csv")

# CHECK MISSING VALUES

print("===== DATASET INFO ====")
df.info()

print("\n===== MISSING VALUES IN EACH COLUMN ====")
print(df.isnull().sum())

print("\n===== TOTAL MISSING VALUES ====")
print(df.isnull().sum().sum())

print("\n===== ANY MISSING VALUES? ====")
print(df.isnull().values.any())

print("\n===== NON-MISSING VALUES COUNT ====")
print(df.notnull().sum())


# FILL MISSING VALUES


# Fill all missing values with 0
df_constant = df.fillna(0)
print("\n===== CONSTANT VALUE ====")
display(df_constant.head())

# Fill '2024' column with mean
df_mean = df.copy()
df_mean["2024"] = df_mean["2024"].fillna(df_mean["2024"].mean())

print("\n===== MEAN ====")
display(df_mean.head())

# Fill '2024' column with median
df_median = df.copy()
df_median["2024"] = df_median["2024"].fillna(df_median["2024"].median())

print("\n===== MEDIAN ====")
display(df_median.head())

# Fill 'Maker' column with mode
df_mode = df.copy()
df_mode["Maker"] = df_mode["Maker"].fillna(df_mode["Maker"].mode()[0])

print("\n===== MODE ====")
display(df_mode.head())

# Forward Fill
df_ffill = df.ffill()

print("\n===== FORWARD FILL ====")
display(df_ffill.head())

# Backward Fill
df_bfill = df.bfill()

print("\n===== BACKWARD FILL ====")
display(df_bfill.head())



# DROP MISSING VALUES


# Drop rows with missing values
df_drop_rows = df.dropna()

print("\n===== DROP ROWS ====")
display(df_drop_rows.head())

# Drop columns with missing values
df_drop_cols = df.dropna(axis=1)

print("\n===== DROP COLUMNS ====")
display(df_drop_cols.head())

# Drop rows having any missing value
df_any = df.dropna(how="any")

print("\n===== DROP HOW='ANY' ====")
display(df_any.head())

# Drop rows where all values are missing
df_all = df.dropna(how="all")

print("\n===== DROP HOW='ALL' ====")
display(df_all.head())

# Keep rows with at least 5 non-null values
df_thresh = df.dropna(thresh=5)

print("\n===== THRESH = 5 ====")
display(df_thresh.head())

# Drop rows where 'Maker' or 'Cat' is missing
df_subset = df.dropna(subset=["Maker", "Cat"])

print("\n===== SUBSET (Maker, Cat) ====")
display(df_subset.head())

# Drop missing values permanently
df_copy = df.copy()
df_copy.dropna(inplace=True)

print("\n===== AFTER inplace=True ====")
display(df_copy.head())
```
