```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("ev_sales_by_makers_and_cat_15-24.csv")

print("Shape:", df.shape)
```

```python
years = [
    "2015", "2016", "2017", "2018", "2019",
    "2020", "2021", "2022", "2023", "2024"
]

yearly_sales = df[years].sum()

plt.figure(figsize=(10,5))
plt.plot(years, yearly_sales, marker='o')
plt.title("Year-wise EV Sales")
plt.xlabel("Year")
plt.ylabel("Number of EV Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
category_sales = df.groupby("Cat")["2024"].sum()

plt.figure(figsize=(7,5))
plt.bar(category_sales.index, category_sales.values)
plt.title("EV Sales by Vehicle Category")
plt.xlabel("Vehicle Category")
plt.ylabel("EV Sales in 2024")
plt.show()
```

```python
category_sales = df.groupby("Cat")["2024"].sum()

plt.figure(figsize=(8,6))
plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct='%1.1f%%'
)
plt.title("EV Sales Distribution by Vehicle Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["2024"].dropna(), bins=30)
plt.title("Distribution of EV Sales in 2024")
plt.xlabel("EV Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["2023"],
    df["2024"],
    alpha=0.3
)

plt.title("EV Sales: 2023 vs 2024")
plt.xlabel("EV Sales in 2023")
plt.ylabel("EV Sales in 2024")
plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.boxplot(df["2024"].dropna())

plt.title("EV Sales in 2024 Box Plot")
plt.ylabel("EV Sales")
plt.show()
```

```python
print("Total EV Sales (2015-2024):", df[years].sum().sum())
print("EV Sales in 2024:", df["2024"].sum())
print("Average EV Sales in 2024:", round(df["2024"].mean(), 2))
print("Most Popular Vehicle Category:", df.groupby("Cat")["2024"].sum().idxmax())
print("Top EV Maker in 2024:", df.groupby("Maker")["2024"].sum().idxmax())
```
